# Dome Azimuth Analysis

This notebook is to analyze the azimuth movement of dome.

## Import Modules

This notebook needs to setup the **ts_aos_utils** under the **notebooks/.user_setups**.

```python
%matplotlib inline
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
from astropy.time import Time

from lsst.ts.aos.utils import DiagnosticsMTDome, EfdName
```

## Instantiate the DiagnosticsMTDome Class

```python
diagnostics_dome = DiagnosticsMTDome(efd_name=EfdName.Usdf)
```

## Check the Data

Notice that the UTC time is used when doing the query.

```python
time_start = Time("2025-03-18T12:14:00", scale="utc", format="isot")
time_end = Time("2025-03-18T12:16:44", scale="utc", format="isot")
```

```python
data_azimuth, time_operation = await diagnostics_dome.get_data_azimuth(
    time_start,
    time_end,
    realign_time=True,
)
```

```python
diagnostics_dome.plot_azimuth_position_velocity(data_azimuth, time_operation)
```

```python
diagnostics_dome.plot_azimuth_drive_current(data_azimuth, time_operation)
```
